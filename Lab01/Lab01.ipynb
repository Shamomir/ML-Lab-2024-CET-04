Task 7

```python
def get_domain(email):
    position = email.find("@")
    return email[position + 1:]

traceDomain = get_domain("user@domain.com")
print("Q7:", traceDomain)
```

Task 8

```python
def has_dog(text):
    text = text.lower()
    if "dog" in text:
        return True
    else:
        return False

srchDog = has_dog("The dog barked")
print("Q8:", srchDog)
```

Task 9

```python
def count_dog(text):
    text = text.lower()
    return text.count("dog")

dogCounter = count_dog("Dog dog DOG cat")
print("Q9:", dogCounter)
```

Task 10

```python
def caught_speeding(speed, is_birthday):
    if is_birthday == True:
        if speed <= 65:
            return "No Ticket"
        elif speed <= 85:
            return "Small Ticket"
        else:
            return "Big Ticket"
    else:
        if speed <= 60:
            return "No Ticket"
        elif speed <= 80:
            return "Small Ticket"
        else:
            return "Big Ticket"


OutPut1 = caught_speeding(65, True)
OutPut2 = caught_speeding(85, True)
OutPut3 = caught_speeding(90, False)

print("Q10:", OutPut1, OutPut2, OutPut3)
```
