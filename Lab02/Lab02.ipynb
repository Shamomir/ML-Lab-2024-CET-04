**Task 06**

```python
admin_units = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
pop_17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
pop23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def to_dict(turn_pairs):
    out = {}
    for ki, value in turn_pairs:
        out[str(ki)] = value.item() if hasattr(value, "item") else value
    return out

pop_total = np.sum(pop23)
print(f"T6a: total population 2023 = {pop_total:.2f} million")

percent_share = (pop23 / pop_total) * 100
print(f"T6b: share of total (%): {to_dict(zip(admin_units, np.round(percent_share, 2)))}")

print(f"T6c: most populous unit: {admin_units[pop23.argmax()]}")
print(f"T6d: units above 20 million: {admin_units[pop23 > 20]}")

yearly_growth = 100 * ((pop23 / pop_17) ** (1 / 6) - 1)
print(f"T6e: annual growth 2017-2023 (% per year): {to_dict(zip(admin_units, np.round(yearly_growth, 2)))}")

print(f"T6f: fastest-growing unit: {admin_units[np.argmax(yearly_growth)]} ({yearly_growth.max():.2f}% per year)")

nat = ((pop23.sum() / pop_17.sum()) ** (1 / 6) - 1) * 100
doubling_time = 70 / yearly_growth.max()
print(f"T6 interpret: national growth = {nat:.2f}% per year.")
print(f"T6 interpret: {admin_units[np.argmax(yearly_growth)]} would double in about {doubling_time:.0f} years at its current rate.")
print(f"T6 interpret: AJK grew {yearly_growth[-1]:.2f}% per year vs national {nat:.2f}% -> "
      f"AJK is {nat - yearly_growth[-1]:.2f} points below the national rate (in this sample AJK is essentially flat).")
```

**Task 07**

```python
city_arr = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
month_list = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temp_grid = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rain = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def to_dict(turn_pairs):
    out = {}
    for ki, value in turn_pairs:
        out[str(ki)] = value.item() if hasattr(value, "item") else value
    return out

print(f"T7a: shape of temp (cities, months): {temp_grid.shape}")

annual_mean = np.mean(temp_grid, axis=1)
print(f"T7b: yearly mean temperature (deg C): {to_dict(zip(city_arr, np.round(annual_mean, 1)))}")

peak_month = month_list[np.argmax(temp_grid, axis=1)]
print(f"T7c: hottest month per city: {to_dict(zip(city_arr, peak_month))}")

print(f"T7d: 5-city monthly average (deg C): {to_dict(zip(month_list, np.round(temp_grid.mean(axis=0), 1)))}")

n_above35 = (temp_grid > 35).sum(axis=1)
print(f"T7e: months above 35 deg C: {to_dict(zip(city_arr, n_above35))}")

f_lahore = temp_grid[0] * 9 / 5 + 32
print(f"T7f: Lahore monthly temperature (deg F): {np.round(f_lahore, 1)}")

monsoon = np.sum(rain[:, 6:9], axis=1)
print(f"T7g: monsoon rain Jul-Sep (mm): {to_dict(zip(city_arr, monsoon))}")
print(f"T7g: wettest monsoon city: {city_arr[np.argmax(monsoon)]} ({monsoon.max()} mm)")

percent_share = monsoon / rain.sum(axis=1) * 100
print(f"Task 7h -> monsoon share of annual rain (%): {to_dict(zip(city_arr, np.round(percent_share, 1)))}")

ki = np.flatnonzero(city_arr == "Karachi")[0]
print(f"T7 interpret: Karachi has the highest yearly mean ({annual_mean.max():.1f} deg C) "
      f"but {n_above35[ki]} months above 35 deg C, while cooler-on-average Peshawar has {n_above35[4]} "
      f"and Lahore {n_above35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

**Task 08**

```python
years_fy = np.array(["FY2023", "FY2024", "FY2025"])
month_fy = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

infl = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def to_dict(turn_pairs):
    out = {}
    for ki, value in turn_pairs:
        out[str(ki)] = value.item() if hasattr(value, "item") else value
    return out

avg_infl = np.mean(infl, axis=1)
print(f"T8a: average inflation per year (%): {to_dict(zip(years_fy, np.round(avg_infl, 2)))}")

fy24_vals = infl[1]
print(f"T8b: FY2024 months above 25 %: {month_fy[fy24_vals > 25]} -> {(fy24_vals > 25).sum()} months")

diff_arr = np.diff(infl, axis=1)
print(f"T8c: month-on-month change (percentage points; negative = drop):\n{np.round(diff_arr, 1)}")

big_drop = diff_arr.argmin(axis=1)
print(f"T8d: month of biggest drop: {to_dict(zip(years_fy, month_fy[big_drop + 1]))} | size (pp): {np.round(diff_arr.min(axis=1), 1)}")

fy25 = infl[2]
labels = np.where(fy25 < 5, "Low", np.where(fy25 < 10, "Moderate", "High"))
print(f"T8e: FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((month_fy, fy25, labels))}")

print(f"T8 interpret: average inflation went {avg_infl[0]:.1f}% -> {avg_infl[1]:.1f}% -> {avg_infl[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

**Task 09**

```python
exam_marks = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
roll_ids = np.arange(1, 21)
THRESHOLD = 40

new_marks = np.clip(5 + exam_marks, 0, 100)
print(f"T9a: original marks: {exam_marks}")
print(f"T9a: marks after +5 grace (cap 100): {new_marks}")

pass_before = (exam_marks >= THRESHOLD).mean() * 100
pct_after = (new_marks >= THRESHOLD).mean() * 100
print(f"T9b: pass % before grace = {pass_before:.1f} %, after grace = {pct_after:.1f} %")

grace_mask = (exam_marks < THRESHOLD) & (new_marks >= THRESHOLD)
print(f"T9c: passed only because of grace (roll numbers): {roll_ids[grace_mask]}")
print(f"T9c: their original marks: {exam_marks[grace_mask]}")

zs = (new_marks - np.mean(new_marks)) / np.std(new_marks)
print(f"T9d: mean = {new_marks.mean():.2f}, std = {new_marks.std():.2f}")
print(f"T9d: z-scores (std devs from mean): {np.round(zs, 2)}")
```

**Task 10**

```python
canvas = np.zeros((8, 8), dtype=np.uint8)
canvas[1, 1:-1] = 255
canvas[[2, 3, 4, 5, 6], [6, 5, 5, 4, 4]] = 255

def render(a):
    return (a > 0).astype(int)

print(f"T10a: the '7' (1 = white):\n{render(canvas)}")

img_flip = np.fliplr(canvas)
print(f"T10b: flipped left-right:\n{render(img_flip)}")

inv_img = ~canvas
print(f"T10c: inverted (white <-> black):\n{render(inv_img)}")

sub_img = canvas[0:7, 1:8]
print(f"T10d: cropped 7x7 region, shape: {sub_img.shape}\n{render(sub_img)}")

bright = np.clip(canvas.astype(int) + 60, 0, 255).astype(np.uint8)
print(f"T10e: brightened (+60, capped at 255); unique values: {np.unique(bright)}")

print(f"T10f: shape = {canvas.shape} | dtype = {canvas.dtype} | white pixels = {np.count_nonzero(canvas == 255)}")
```
